```python
import pandas as pd
import numpy as np
import sklearn as sci
import re 
import string
import os

from transformers import AutoTokenizer

#import datasets 
#import torch
```

```python
df = pd.read_csv("../Data/raw/phishing_email.csv")
df.head(500)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python

##
#This file is meant to be the preprocessing file that will
#make the text uniform, remove URLs, HTML tags, punctuation, and numbers
#clean the extra spaces that are within the every email
## 

def clean_text(text):
    text = text.lower() #lowercase 
    text = re.sub(r"http\S+|www\S+", "", text) #url removal
    text = re.sub(r"<.*?>", "", text) #http removal
    text = re.sub(f" [{string.punctuation}]", "", text) #punctuation removal
    text = re.sub(r"\d+", "", text) #number removal
    text = text.strip() #remove spaces 
    return text

df["Cleaned_text"] = df["text_combined"].apply(clean_text)

print(df[["text_combined", "Cleaned_text"]].head())
```

```python
## Start loading BERT
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
#use a sample email
sample_email = df["Cleaned_text"][0]
tokens = tokenizer.tokenize(sample_email)
token_ids = tokenizer.convert_tokens_to_ids(tokens)
```

```python
print("Original Text:", sample_email)
print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

```python
df[["Cleaned_text", "label"]].to_csv("../Data/processed/phishing_email_data.csv", index=False)
print("Data is Cleaned and saved")
```
