```python
import pickle
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
from pathlib import Path

file_path = Path("../Data/processed/processed_data.pkl")

print("File size:", file_path.stat().st_size, "bytes")
print("First 150 bytes:")
print(file_path.read_bytes()[:150])
```

```python
with open("../Data/processed/processed_data.pkl", "rb") as f:
    X_train, X_test, train_labels, test_labels, vectorizer = pickle.load(f)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
print("Training labels shape:", train_labels.shape)
print("Testing labels shape:", test_labels.shape)
```

### Step 2: Train a Logistic Regression Model
This is a common classification that should work well with text data

```python
print("X_train shape:", X_train.shape)
print("Train labels shape:", train_labels.shape)

print("X_test shape:", X_test.shape)
print("Test labels shape:", test_labels.shape)

print("Training difference:", len(train_labels) - X_train.shape[0])
print("Testing difference:", len(test_labels) - X_test.shape[0])
```

```python

print("Training label distribution:")
print(pd.Series(train_labels).value_counts(dropna=False))

print("\nTesting label distribution:")
print(pd.Series(test_labels).value_counts(dropna=False))
```

```python
print("Training labels type:", type(train_labels))
print("Training labels index:")
print(train_labels.index[:10])

print("\nLast 10 training labels:")
print(train_labels.tail(10))
```

```python
# Initialize the model
model = LogisticRegression(max_iter=1000)

# Train the model on training data
model.fit(X_train, train_labels)

print("Model training complete!")
```

### Step 3: Evaluate Model Performance

```python
predictions = model.predict(X_test)
```

```python
# Calculate evaluation metrics
accuracy = accuracy_score(test_labels, predictions)
precision = precision_score(test_labels, predictions)
recall = recall_score(test_labels, predictions)
f1 = f1_score(test_labels, predictions)

# Print results
print(f"Accuracy: {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1-score: {f1:.2f}")
```

```python
# Define the correct folder relative to the notebook
model_path = Path("../Data/processed/phishing_model.pkl")

# Ensure the destination folder exists
model_path.parent.mkdir(parents=True, exist_ok=True)

# Save the trained model
with open(model_path, "wb") as f:
    pickle.dump(model, f)

print("Trained model saved!")
print("Model location:", model_path.resolve())
```
