```python
import pickle
import numpy as np

from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
# Load vectorized data and labels
with open("../Data/processed/processed_data.pkl", "rb") as f:
    X_train, X_test, train_labels, test_labels, vectorizer = pickle.load(f)

# Load trained model
with open("../Data/processed/phishing_model.pkl", "rb") as f:
    model = pickle.load(f)

print("Processed data loaded successfully!")
print("Trained model loaded successfully!")
```

### Step 2: Improve Logistic Regression with Hyperparameter Tuning

```python
# Define parameter grid
param_grid = {
    "C": [0.01, 0.1, 1, 10, 100],  # Regularization strength
    "max_iter": [500, 1000, 2000]  # Number of training iterations
}

# Initialize Grid Search with cross-validation
grid_search = GridSearchCV(LogisticRegression(), param_grid, cv=5, scoring="accuracy", verbose=1)

# Train on dataset
grid_search.fit(X_train, train_labels)

# Get best model
best_model = grid_search.best_estimator_

# Display best parameters
print("Best Parameters:", grid_search.best_params_)
```

### Step 3: Try Alternative Models
SVM and Random Forest Model

```python
# Initialize and train SVM
svm_model = SVC(kernel="linear", C=1)
svm_model.fit(X_train, train_labels)

# Predict and evaluate
svm_predictions = svm_model.predict(X_test)
```

```python
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, train_labels)

# Predict and evaluate
rf_predictions = rf_model.predict(X_test)
```

### Step 4: Evaluate and Compare All the Models
This is comparing the Logistic Regression, SVM, and Random Forest

```python
# Logistic Regression
lr_accuracy = accuracy_score(test_labels, best_model.predict(X_test))
print(f"Logistic Regression Accuracy: {lr_accuracy:.2f}")
lr_precision = precision_score(test_labels, best_model.predict(X_test))
print(f"Logistic Regression Precision: {lr_precision:.2f}")
lr_recall = recall_score(test_labels, best_model.predict(X_test))
print(f"Logistic Regression Recall: {lr_recall:.2f}")
lr_f1_score = f1_score(test_labels, best_model.predict(X_test))
print(f"Logistic Regression F1 Score: {lr_f1_score:.2f}")

# SVM
svm_accuracy = accuracy_score(test_labels, svm_predictions)
print(f"SVM Accuracy: {svm_accuracy:.2f}")
svm_precision = precision_score(test_labels, svm_predictions)
print(f"SVM Precision: {svm_precision:.2f}")
svm_recall = recall_score(test_labels, svm_predictions)
print(f"SVM Recall: {svm_recall:.2f}")
svm_f1_score = f1_score(test_labels, svm_predictions)
print(f"SVM F1 Score: {svm_f1_score}")

# Random Forest
rf_accuracy = accuracy_score(test_labels, rf_predictions)
print(f"Random Forest Accuracy: {rf_accuracy:.2f}")
rf_precision = precision_score(test_labels, rf_predictions)
print(f"Random Forest Precision: {rf_precision:.2f}")
rf_recall = recall_score(test_labels, rf_predictions)
print(f"Random Forest Recall: {rf_recall}")
rf_f1_score = f1_score(test_labels, rf_predictions)
print(f"Random Forest F1 Score: {rf_f1_score}")
```
