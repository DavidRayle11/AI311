```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
import pickle
```

```python
df = pd.read_csv("../Data/processed/phishing_email_data_main.csv")
df.head(500)
```

```python
print(df["label"].value_counts())
```

### Convert Labels

```python
df.head()
```

```python
df.dtypes
```

```python
train_text, test_text, train_labels, test_labels = train_test_split(
    df["Cleaned_text"], df["label"], 
    test_size=0.2, random_state=42
)
```

```python
print(f"Training set size: {len(train_text)}")
print(f"Testing set size: {len(test_text)}")
```

```python
# initialize the vectorizer 
vect = TfidfVectorizer(max_features=5000)
```

```python
print(train_text.isna().sum())
print(test_text.isna().sum())
```

```python
train_text.dropna(inplace=True)
```

```python
X_train = vect.fit_transform(train_text)
X_test = vect.transform(test_text)
```

```python
print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
with open("../Data/processed/processed_data.pkl", "wb") as f:
    pickle.dump((X_train, X_test, train_labels, test_labels, vect), f)

print("Processed data saved to processed_data.pkl")
```
