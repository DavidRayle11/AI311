```python
import pickle
import numpy as np
```

```python
# Load the trained model
with open("../Data/processed/best_phishing_model.pkl", "rb") as f:
    model = pickle.load(f)

# Load the vectorizer
with open("../Data/processed/processed_data.pkl", "rb") as f:
    _, _, _, _, vectorizer = pickle.load(f)

print("Model and vectorizer loaded successfully!")

```

### Step 2: Define A Function to Classify Emails
This will help us take an email as input

```python
def classify_email(email_text):
    """Classifies an email as phishing or legitimate."""
    # Convert email text into the format the model understands
    email_vectorized = vectorizer.transform([email_text])

    # Make prediction
    prediction = model.predict(email_vectorized)

    # Interpret result
    result = "Phishing Email" if prediction[0] == 1 else "Legitimate Email"
    return result

# Example test
test_email = "Click this link to reset your password: http://fakebank.com"
print("Test Email Classification:", classify_email(test_email))
```

### Step 3: Build a Command-Line Tool
This will be an interactive script where users enter email text and get predictions

```python
if __name__ == "__main__":
    print("🔍 Phishing Detection Tool 🔍")
    print("Type or paste an email message below and press Enter.")
    print("Type 'exit' to quit.\n")

    while True:
        user_input = input("Enter email text: ")

        # Exit condition
        if user_input.lower() == "exit":
            print("Goodbye!")
            break

        # Classify the email
        print("Result:", classify_email(user_input))
        print("\n")
```

```python
with open("../Data/processed/final_phishing_model.pkl", "wb") as f:
    pickle.dump(model, f)

print("Final model saved for future use!")
```
