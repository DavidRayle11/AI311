# Data Quality Scorecard

## Completeness

```python
total_cells = customer.shape[0] * customer.shape[1]
filled_cells = customer.notna().sum().sum()
completeness = (filled_cells / total_cells) * 100
print(round(completeness, 1))
```

## Duplicate Percentage

```python
duplicate_percentage = (customer.duplicated().sum() / len(customer)) * 100
print(round(duplicate_percentage, 1))
```

## Age Validity

```python
valid_age_percentage = (customer["Age"].between(18, 100).sum() / len(customer)) * 100
print(round(valid_age_percentage, 1))
```

## Metric Consistency

```python
standard_states = ["Tennessee", "Georgia", "New York", "Illinois",
                   "Texas", "Florida", "California", "North Carolina"]
state_consistency = (customer["State"].isin(standard_states).sum() / customer["State"].notna().sum()) * 100
print(round(state_consistency, 1))
```

## Data Cleaning

```python
clean = customer.copy()
```

## Remove Exact Duplicates

```python
clean = clean.drop_duplicates().copy()
print("Rows after duplicate removal:", len(clean))
```

## Standardize Names

```python
state_mapping = {
    "TN":"Tennessee", "Tennesee":"Tennessee", "tennessee":"Tennessee",
    "GA":"Georgia", "georgia":"Georgia",
    "NY":"New York", "new york":"New York",
    "IL":"Illinois", "TX":"Texas", "FL":"Florida",
    "CA":"California", "NC":"North Carolina",
    "N. Carolina":"North Carolina"
}
clean["State"] = clean["State"].replace(state_mapping)
print(clean["State"].value_counts(dropna=False).sort_index())
```

**Question 10** Why is standardizing values usually better than simply deleting every row with an inconsistent state name?

 The data still has meaning, the records are still real records of real people but they mispelled something, that does not mean that their record should get deleted. While deleting may be easier in the short term there are a lot of vulnerabilities that could be overlooked in deleting even one of the rows. 

## Correct Invalid Ages and Missing Ages

**Question 11** What median age was used? Explain one advantage and one limitation of filling missing ages with the median.

The median age was 44. The median is useful because it not really affected much by outliers. The main limitation about this is that many missing ages with the same median value can reduce the natural variation amongst the ages in the data.

```python
clean.loc[~clean["Age"].between(18, 100), "Age"] = pd.NA
median_age = clean["Age"].median()
clean["Age"] = clean["Age"].fillna(median_age)
print("Median age used:", median_age)
print(clean["Age"].describe())
```
